```python
import numpy as np
import pandas as pd

np.random.seed(42)

bill_ids = np.arange(1001, 1301)

products = [
    "Rice",
    "Wheat Flour",
    "Sugar",
    "Milk",
    "Cooking Oil",
    "Tea",
    "Biscuits",
    "Salt"
]

bill_dates = []

for i in range(300):
    d = np.random.randint(1, 29)
    m = np.random.randint(1, 4)

    bill_dates.append(f"{d:02d}/{m:02d}/2026")

sales_data = {
    "bill_id": bill_ids,
    "bill_date": bill_dates,
    "product": np.random.choice(products, 300),
    "qty": np.random.randint(1, 11, 300),
    "unit_price_inr": np.random.randint(10, 501, 300),
    "payment_mode": np.random.choice(
        ["Cash", "UPI", "Card"], 300
    )
}

sales_df = pd.DataFrame(sales_data)

sales_df.to_csv("kirana_sales.csv", index=False)

sales_df.head()
```
